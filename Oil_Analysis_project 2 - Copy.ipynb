## 🧹 Data Cleaning

This step handles missing values and prepares the dataset for analysis.

```python
# Function to clean data
def clean_data(df):
    """
    This function handles missing values by replacing them with column means.
    """
    df = df.copy()
    
    cols = [
        'AVG_DOWNHOLE_PRESSURE',
        'AVG_DOWNHOLE_TEMPERATURE',
        'AVG_ANNULUS_PRESS',
        'AVG_WHP_P',
        'AVG_WHT_P'
    ]
    
    for col in cols:
        # Fill missing values with mean
        df[col].fillna(df[col].mean(), inplace=True)
    
    return df

# Apply cleaning
df = clean_data(df)
```

## ⚙️ Feature Engineering

We create new features to better understand production behavior.

```python
# Function to create new features
def create_features(df):
    """
    This function creates new engineered features such as total liquid and water cut.
    """
    df = df.copy()
    
    # Total liquid production
    df['TOTAL_LIQUID'] = df['BORE_OIL_VOL'] + df['BORE_WAT_VOL']
    
    # Water cut ratio
    df['WATER_CUT'] = df['BORE_WAT_VOL'] / df['TOTAL_LIQUID']
    
    return df

df = create_features(df)
```
